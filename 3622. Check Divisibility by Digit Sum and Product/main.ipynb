```python
class Solution:
  def checkDivisibility(self, n: int) -> bool:
    digit_sum = 0
    digit_product = 1
    for char in str(n):
      digit_sum += int(char)
      digit_product *= int(char)
    return n % (digit_sum + digit_product) == 0
```

```python
n = 99
Solution().checkDivisibility(n=n)
```

```python
n = 23
Solution().checkDivisibility(n=n)
```
