```python
class Solution:
  def evaluate(self, s: str, knowledge: list[list[str]]) -> str:
    d = dict(knowledge)
    ans, start = [], -1
    for i, c in enumerate(s):
      if c == '(':
        start = i
      elif c == ')':
        ans.append(d.get(s[start + 1: i], '?'))
        start = -1
      elif start < 0:
        ans.append(c)
    return ''.join(ans)
```

```python
s = "(name)is(age)yearsold"
knowledge = [["name", "bob"], ["age", "two"]]
Solution().evaluate(s=s, knowledge=knowledge)
```

```python
s = "hi(name)"
knowledge = [["a", "b"]]
Solution().evaluate(s=s, knowledge=knowledge)
```

```python
s = "(a)(a)(a)aaa"
knowledge = [["a", "yes"]]
Solution().evaluate(s=s, knowledge=knowledge)
```
