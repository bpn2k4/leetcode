```python
from typing import List


class Solution:
  def climbStairs(self, n: int, costs: List[int]) -> int:
    dp = [0] * (n + 1)
    for i in range(1, n + 1):
      dp[i] = min(
        dp[i - 1] + 1, 
        dp[i - 2] + 2 * 2, 
        dp[i - 3] + 3 * 3,
      ) + costs[i - 1]
    return dp[n]
```

```python
n = 4
costs = [1, 2, 3, 4]
Solution().climbStairs(n=n, costs=costs)
```
