```python
from typing import Tuple


class Solution:
  def sumGame(self, num: str) -> bool:

    def get(s: str) -> Tuple[int, int]:
      nn = qq = 0
      for ch in s:
        if ch == '?':
          qq += 1
        else:
          nn += int(ch)
      return nn, qq

    n = len(num)
    n0, q0 = get(num[: n // 2])
    n1, q1 = get(num[n // 2:])
    return (q0 + q1) % 2 == 1 or n0 - n1 != (q1 - q0) * 9 // 2
```

```python
num = "5023"
Solution().sumGame(num=num)
```
