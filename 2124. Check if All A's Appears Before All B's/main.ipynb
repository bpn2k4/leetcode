```python
class Solution:
  def checkString(self, s: str) -> bool:
    last_a = s.rfind('a')
    first_b = s.find('b')
    if last_a == -1 or first_b == -1:
      return True
    return last_a < first_b
```

```python
s = "aaabbb"
Solution().checkString(s=s)
```

```python
s = "abab"
Solution().checkString(s=s)
```

```python
s = "bbb"
Solution().checkString(s=s)
```
