```python
from typing import List


class Solution:
  def sumOfSquares(self, nums: List[int]) -> int:
    n = len(nums)
    s = 0
    for index, num in enumerate(nums, start=1):
      if n % index == 0:
        s += num * num
    return s
```

```python
nums = [1, 2, 3, 4]
Solution().sumOfSquares(nums=nums)
```

```python
nums = [2, 7, 1, 19, 18, 3]
Solution().sumOfSquares(nums=nums)
```
