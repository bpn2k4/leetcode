```python
from typing import List


class Solution:
  def missingMultiple(self, nums: List[int], k: int) -> int:
    nums = set(nums)
    num = k
    while num in nums:
      num += k
    return num
```

```python
nums = [8, 2, 3, 4, 6]
k = 2
Solution().missingMultiple(nums=nums, k=k)
```

```python
nums = [1, 4, 7, 10, 15]
k = 5
Solution().missingMultiple(nums=nums, k=k)
```
