```python
class Solution:
  def firstStableIndex(self, nums: list[int], k: int) -> int:
    n = len(nums)
    max_of_index = [nums[0]]
    for i in range(1, n):
      if nums[i] > max_of_index[-1]:
        max_of_index.append(nums[i])
      else:
        max_of_index.append(max_of_index[-1])
    min_of_index = [nums[-1]]
    for i in range(n - 2, -1, -1):
      if nums[i] < min_of_index[-1]:
        min_of_index.append(nums[i])
      else:
        min_of_index.append(min_of_index[-1])
    for i in range(n):
      m = max_of_index[i] - min_of_index[n - 1 - i]
      if m <= k:
        return i
    return -1
  
  def firstStableIndex(self, nums: list[int], k: int) -> int:
    n = len(nums)
    min_of_index = [nums[-1]]
    for i in range(n - 2, -1, -1):
      if nums[i] < min_of_index[-1]:
        min_of_index.append(nums[i])
      else:
        min_of_index.append(min_of_index[-1])
    max_num = -1
    for i in range(n):
      max_num = max(max_num, nums[i])
      m = max_num - min_of_index[n - 1 - i]
      if m <= k:
        return i
    return -1
```
