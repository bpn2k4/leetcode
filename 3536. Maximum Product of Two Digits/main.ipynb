```python
class Solution:
  def maxProduct(self, n: int) -> int:
    digit = [int(i) for i in str(n)]
    max_product = 0
    for i in range(len(digit)):
      for j in range(i + 1, len(digit)):
        max_product = max(max_product, digit[i] * digit[j])
    return max_product
```

```python
n = 31
Solution().maxProduct(n=n)
```
