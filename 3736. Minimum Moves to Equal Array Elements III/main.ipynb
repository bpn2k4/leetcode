```python
from typing import List


class Solution:
  def minMoves(self, nums: List[int]) -> int:
    max_num = max(nums)
    count = 0
    for num in nums:
      count += max_num - num
    return count
```

```python
nums = [2, 1, 3]
Solution().minMoves(nums=nums)
```

```python
nums = [4, 4, 5]
Solution().minMoves(nums=nums)
```
