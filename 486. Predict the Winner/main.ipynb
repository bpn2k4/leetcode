```python
from typing import List


class Solution:
  def predictTheWinner(self, nums: List[int]) -> bool:
    n = len(nums)
    dp = [0] * n
    for i in range(n - 1, -1, -1):
      dp[i] = nums[i]
      for j in range(i + 1, n):
        dp[j] = max(nums[i] - dp[j], nums[j] - dp[j - 1])
    return dp[n - 1] >= 0
```

```python
nums = [1, 5, 2]
Solution().predictTheWinner(nums=nums)
```

```python
nums = [1, 5, 233, 7]
Solution().predictTheWinner(nums=nums)
```
