```python
from typing import List


class Solution:
  def minSumOfLengths(self, arr: List[int], target: int) -> int:
    n = len(arr)
    ans = n + 1
    total = 0
    dp = [n] * (n + 1)
    left = 0
    for right, num in enumerate(arr):
      total += num
      while total > target:
        total -= arr[left]
        left += 1
      dp[right + 1] = dp[right]
      if total == target:
        ans = min(ans, right - left + 1 + dp[left])
        dp[right + 1] = min(dp[right], right - left + 1)
    if ans == n + 1:
      return -1
    return ans
```

```python
arr = [3, 2, 2, 4, 3]
target = 3
Solution().minSumOfLengths(arr=arr, target=target)
```
