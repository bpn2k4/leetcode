```python
from functools import cache
import math


class Solution:
  def winnerSquareGame(self, n: int) -> bool:

    @cache
    def dfs(i: int) -> bool:
      if i == 0:
        return False
      for j in range(1, int(math.sqrt(i)) + 1):
        if not dfs(i - j * j):
          return True
      return False

    return dfs(n)
```

```python
n = 1
Solution().winnerSquareGame(n=n)
```
