```python
class Solution:
  def minMoves2(self, nums: list[int]) -> int:
    n = len(nums)
    nums.sort()
    median = nums[n // 2]
    count = 0
    for num in nums:
      count += abs(num - median)
    return count
```

```python
nums = [1, 2, 3]
Solution().minMoves2(nums=nums)
```

```python
nums = [1, 10, 2, 9]
Solution().minMoves2(nums=nums)
```
