```python
from typing import List
from collections import Counter


class Solution:
  def largestInteger(self, nums: List[int], k: int) -> int:
    if k == len(nums):
      return max(nums)
    num2count = Counter(nums)
    if k == 1:
      m = max([num for num, count in num2count.items() if count == 1] + [-1])
      if m != -1:
        return m
    first = nums[0]
    last = nums[-1]
    if num2count[first] > 1:
      first = -1
    if num2count[last] > 1:
      last = -1
    return max(first, last, -1)
```

```python
nums = [3, 9, 2, 1, 7]
k = 3
Solution().largestInteger(nums=nums, k=k)
```

```python
nums = [3, 9, 7, 2, 1, 7]
k = 4
Solution().largestInteger(nums=nums, k=k)
```
