```python
class Solution:
  def maxSumOfSquares(self, num: int, sum: int) -> str:
    if 9 * num < sum:
      return ''
    c = sum // 9
    r = sum % 9
    n = c * '9'
    if r != 0:
      n += str(r)
    n += '0' * (num -len(n))
    return n
```

```python
num = 2
sum = 3
Solution().maxSumOfSquares(num=num, sum=sum)
```
