```python
class Solution:
  def scoreOfParentheses(self, s: str) -> int:
    ans = bal = 0
    for i, char in enumerate(s):
      if char == '(':
        bal += 1
      else:
        bal -= 1
        if s[i - 1] == '(':
          ans += 1 << bal
    return ans
```

```python
s = "()"
Solution().scoreOfParentheses(s=s)
```

```python
s = "(())"
Solution().scoreOfParentheses(s=s)
```
