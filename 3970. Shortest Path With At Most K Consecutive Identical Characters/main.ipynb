```python
from typing import List
import heapq


class Solution:
  def shortestPath(self, n: int, edges: List[List[int]], labels: str, k: int) -> int:
    g = [[] for _ in range(n)]
    for x, y, w in edges:
      g[x].append((y, w))
    dis = [[2**31 - 1] * (k + 1) for _ in range(n)]
    pq = []
    dis[0][1] = 0
    heapq.heappush(pq, (0, 0, 1))
    while pq:
      d, x, cnt = heapq.heappop(pq)
      if d > dis[x][cnt]:
        continue
      if x == n - 1:
        return d
      for y, w in g[x]:
        new_cnt = cnt + 1 if labels[x] == labels[y] else 1
        if new_cnt > k:
          continue
        nd = d + w
        if nd < dis[y][new_cnt]:
          dis[y][new_cnt] = nd
          heapq.heappush(pq, (nd, y, new_cnt))
    return -1
```

```python
n = 3
edges = [[0, 1, 1], [1, 2, 1], [0, 2, 3]]
labels = "aab"
k = 1
Solution().shortestPath(n=n, edges=edges, labels=labels, k=k)
```
