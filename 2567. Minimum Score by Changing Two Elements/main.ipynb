```python
from typing import List


class Solution:
  def minimizeSum(self, nums: List[int]) -> int:
    n = len(nums)
    nums.sort()
    return min(
      nums[n - 1] - nums[2],
      nums[n - 2] - nums[1],
      nums[n - 3] - nums[0],
    )
```

```python
nums = [1, 4, 7, 8, 5]
Solution().minimizeSum(nums=nums)
```

```python
nums = [1, 4, 3]
Solution().minimizeSum(nums=nums)
```
