```python
class Solution:
  def maxDepth(self, s: str) -> int:
    depth = 0
    max_depth = 0
    for char in s:
      if char == '(':
        depth += 1
        max_depth = max(max_depth, depth)
      elif char == ')':
        depth -= 1
    return max_depth
```

```python
s = "(1+(2*3)+((8)/4))+1"
Solution().maxDepth(s=s)
```

```python
s = "(1)+((2))+(((3)))"
Solution().maxDepth(s=s)
```

```python
s = "()(())((()()))"
Solution().maxDepth(s=s)
```
