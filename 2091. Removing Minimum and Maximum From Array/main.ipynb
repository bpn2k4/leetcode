```python
from typing import List


class Solution:
  def minimumDeletions(self, nums: List[int]) -> int:
    n = len(nums)
    if n <= 2:
      return n
    max_num = max(nums)
    max_idx = nums.index(max_num)
    min_num = min(nums)
    min_idx = nums.index(min_num)
    if max_idx < min_idx:
      min_idx, max_idx = max_idx, min_idx
    return min(
      min_idx + 1 + n - max_idx,
      max_idx + 1,
      n - min_idx,
    )
```

```python
nums = [2, 10, 7, 5, 4, 1, 8, 6]
Solution().minimumDeletions(nums=nums)
```

```python
nums = [0, -4, 19, 1, 8, -2, -3, 5]
Solution().minimumDeletions(nums=nums)
```

```python
nums = [101]
Solution().minimumDeletions(nums=nums)
```
