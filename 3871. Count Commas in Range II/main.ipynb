```python
class Solution:
  def countCommas(self, n: int) -> int:
    if n < 1000:
      return 0
    count = 0
    if 1_000 <= n < 1_000_000:
      count += n - 999
    if 1_000_000 <= n < 1_000_000_000:
      count += 999000 + (n - 999999) * 2
    if 1_000_000_000 <= n < 1_000_000_000_000:
      count += 999000 + 999000000 * 2 + (n - 999999999) * 3
    if 1_000_000_000_000 <= n < 1_000_000_000_000_000:
      count += 999000 + 999000000 * 2 + 999000000000 * 3 + (n - 999999999999) * 4
    if n == 1_000_000_000_000_000:
      count += 3998998998999005
    return count
```

```python
n = 1004590
Solution().countCommas(n=n)
```
