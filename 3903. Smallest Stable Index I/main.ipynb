```python
class Solution:
  def firstStableIndex(self, nums: list[int], k: int) -> int:
    for i in range(len(nums)):
      m = max(nums[:i + 1]) - min(nums[i:])
      if m <= k:
        return i
    return -1
```

```python
nums = [5, 0, 1, 4]
k = 3
Solution().firstStableIndex(nums=nums, k=k)
```

```python
nums = [3, 2, 1]
k = 1
Solution().firstStableIndex(nums=nums, k=k)
```

```python
nums = [0]
k = 0
Solution().firstStableIndex(nums=nums, k=k)
```
