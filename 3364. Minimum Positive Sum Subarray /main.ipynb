```python
from typing import List


class Solution:
  def minimumSumSubarray(self, nums: List[int], l: int, r: int) -> int:
    n = len(nums)
    min_sum = 100000
    for k in range(l, r + 1):
      prefix = 0
      for i in range(k):
        prefix += nums[i]
      if prefix > 0:
        min_sum = min(min_sum, prefix)
      for j in range(k, n):
        prefix += nums[j] - nums[j - k]
        if prefix > 0:
          min_sum = min(min_sum, prefix)
    if min_sum == 100000:
      return -1
    return min_sum
```

```python
nums = [3, -2, 1, 4]
l = 2
r = 3
Solution().minimumSumSubarray(nums=nums, l=l, r=r)
```
