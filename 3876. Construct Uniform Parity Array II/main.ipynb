```python
class Solution:
  def uniformArray(self, nums1: list[int]) -> bool:
    min_num = min(nums1)
    has_odd = any(num % 2 == 1 for num in nums1)
    if min_num % 2 == 1:
      return True
    return not has_odd
```

```python
nums1 = [1, 4, 7]
Solution().uniformArray(nums1=nums1)
```

```python
nums1 = [2, 3]
Solution().uniformArray(nums1=nums1)
```
