```python
class Solution:
  def maximumLengthSubstring(self, s: str) -> int:
    n = len(s)
    max_length = 0
    for left in range(n):
      mark = {s[left]: 1}
      for right in range(left + 1, n):
        char = s[right]
        if mark.get(char, 0) == 2:
          max_length = max(max_length, right - left)
          break
        else:
          max_length = max(max_length, right - left + 1)
        mark[char] = mark.get(char, 0) + 1
    return max_length
```

```python
s = "bcbbbcba"
Solution().maximumLengthSubstring(s=s)
```

```python
s = "aaaa"
Solution().maximumLengthSubstring(s=s)
```
