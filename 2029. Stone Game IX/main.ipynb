```python
from typing import List


class Solution:
  def stoneGameIX(self, stones: List[int]) -> bool:
    r0 = r1 = r2 = 0
    for stone in stones:
      if stone % 3 == 0:
        r0 += 1
      elif stone % 3 == 1:
        r1 += 1
      else:
        r2 += 1
    if r0 % 2 == 0:
      return r1 >= 1 and r2 >= 1
    if r1 - r2 > 2 or r2 - r1 > 2:
      return True
    return False
```

```python
stones = [2, 1]
Solution().stoneGameIX(stones=stones)
```
