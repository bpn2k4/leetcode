```python
from typing import List


class Solution:
  def totalNumbers(self, digits: List[int]) -> int:
    nums = set()
    n = len(digits)
    for i in range(n):
      for j in range(i + 1, n):
        for k in range(j + 1, n):
          n1 = 100 * digits[i] + 10 * digits[j] + digits[k]
          n2 = 100 * digits[k] + 10 * digits[i] + digits[j]
          n3 = 100 * digits[j] + 10 * digits[k] + digits[i]
          n4 = 100 * digits[k] + 10 * digits[j] + digits[i]
          n5 = 100 * digits[i] + 10 * digits[k] + digits[j]
          n6 = 100 * digits[j] + 10 * digits[i] + digits[k]
          if n1 >= 100 and n1 % 2 == 0: nums.add(n1)
          if n2 >= 100 and n2 % 2 == 0: nums.add(n2)
          if n3 >= 100 and n3 % 2 == 0: nums.add(n3)
          if n4 >= 100 and n4 % 2 == 0: nums.add(n4)
          if n5 >= 100 and n5 % 2 == 0: nums.add(n5)
          if n6 >= 100 and n6 % 2 == 0: nums.add(n6)
    return len(nums)
```

```python
digits = [1, 2, 3, 4]
Solution().totalNumbers(digits=digits)
```
