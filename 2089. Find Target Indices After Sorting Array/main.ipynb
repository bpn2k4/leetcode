```python
class Solution:
  def targetIndices(self, nums: list[int], target: int) -> list[int]:
    nums.sort()
    arr = []
    for i in range(len(nums)):
      if nums[i] == target:
        arr.append(i)
    return arr
```

```python
nums = [1, 2, 5, 2, 3]
target = 2
Solution().targetIndices(nums=nums, target=target)
```
