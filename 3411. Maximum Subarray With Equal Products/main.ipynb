```python
from typing import List
import math


class Solution:
  def maxLength(self, nums: List[int]) -> int:
    n = len(nums)
    max_length = 0
    for i in range(n):
      p, g, l = 1, nums[i], nums[i]
      for j in range(i, n):
        g = math.gcd(g, nums[j])
        l = math.lcm(l, nums[j])
        p = p * nums[j]
        if p == g * l:
          max_length = max(max_length, j - i + 1)
    return max_length
```

```python
nums = [1, 2, 1, 2, 1, 1, 1]
Solution().maxLength(nums=nums)
```
