```python
class Solution:
  def minInsertions(self, s: str) -> int:
    count = 0
    need = 0
    for char in s:
      if char == '(':
        if need % 2 == 1:
          count += 1
          need -= 1
        need += 2
      else:
        need -= 1
        if need < 0:
          count += 1
          need = 1
    return count + need
```

```python
s = "(()))"
Solution().minInsertions(s=s)
```
