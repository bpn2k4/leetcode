```python
class Solution:
  def maxDepthAfterSplit(self, seq: str) -> list[int]:
    ans = []
    d = 0
    for c in seq:
      if c == '(':
        d += 1
        ans.append(d % 2)
      if c == ')':
        ans.append(d % 2)
        d -= 1
    return ans
```

```python
seq = "(()())"
Solution().maxDepthAfterSplit(seq=seq)
```
