```python
from typing import List


class Solution:
  def sumCounts(self, nums: List[int]) -> int:
    n = len(nums)
    s = 0
    for i in range(n):
      for j in range(i + 1, n + 1):
        c = len(set(nums[i:j]))
        s += c * c
    return s
```

```python
nums = [1, 2, 1]
Solution().sumCounts(nums=nums)
```

```python
nums = [1, 1]
Solution().sumCounts(nums=nums)
```
