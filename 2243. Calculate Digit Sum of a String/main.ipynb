```python
class Solution:
  def digitSum(self, s: str, k: int) -> str:
    while len(s) > k:
      new_s = ''
      c = 0
      for i in range(len(s)):
        c += int(s[i])
        if (i + 1) % k == 0:
          new_s += str(c)
          c = 0
      if len(s) % k != 0:
        new_s += str(c)
      s = new_s
    return s
```

```python
s = "11111222223"
k = 3
Solution().digitSum(s=s, k=k)
```

```python
s = "00000000"
k = 3
Solution().digitSum(s=s, k=k)
```
