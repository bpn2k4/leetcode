```python
from typing import List


class Solution:
  def longestSubsequence(self, nums: List[int]) -> int:
    n = len(nums)
    total_xor = 0
    all_zero = True
    for x in nums:
      total_xor ^= x
      if x > 0:
        all_zero = False
    if total_xor > 0:
      return n
    if all_zero:
      return 0
    return n - 1
```

```python
nums = [1, 2, 3]
Solution().longestSubsequence(nums=nums)
```
