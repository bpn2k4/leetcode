```python
class Solution:
  def smallestNumber(self, n: int, t: int) -> int:
    num = n
    while True:
      product = 1
      for digit in str(num):
        product *= int(digit)
      if product % t == 0:
        return num
      num += 1
```

```python
n = 10
t = 2
Solution().smallestNumber(n=n, t=t)
```

```python
n = 15
t = 3
Solution().smallestNumber(n=n, t=t)
```
