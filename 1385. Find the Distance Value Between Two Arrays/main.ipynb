```python
class Solution:
  def findTheDistanceValue(self, arr1: list[int], arr2: list[int], d: int) -> int:
    count = 0
    for num1 in arr1:
      valid = True
      for num2 in arr2:
        if abs(num1 - num2) <= d:
          valid = False
          break
      if valid:
        count += 1
    return count
```

```python
arr1 = [4, 5, 8]
arr2 = [10, 9, 1, 8]
d = 2
Solution().findTheDistanceValue(arr1=arr1, arr2=arr2, d=d)
```

```python
arr1 = [1, 4, 2, 3]
arr2 = [-4, -3, 6, 10, 20, 30]
d = 3
Solution().findTheDistanceValue(arr1=arr1, arr2=arr2, d=d)
```
