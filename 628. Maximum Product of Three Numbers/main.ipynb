```python
from typing import List


class Solution:
  def maximumProduct(self, nums: List[int]) -> int:
    nums.sort()
    return max(
      nums[0] * nums[1] * nums[-1],
      nums[-1] * nums[-2] * nums[-3],
    )
  
  def maximumProduct(self, nums: List[int]) -> int:
    max_1 = max_2 = max_3 = float('-inf')
    min_1 = min_2 = float('inf')
    for num in nums:
      if num > max_1:
        max_3 = max_2
        max_2 = max_1
        max_1 = num
      elif num > max_2:
        max_3 = max_2
        max_2 = num
      elif num > max_3:
        max_3 = num
      if num < min_1:
        min_2 = min_1
        min_1 = num
      elif num < min_2:
        min_2 = num
    return max(max_1 * max_2 * max_3, min_1 * min_2 * max_1)
```

```python
nums = [1, 2, 3]
Solution().maximumProduct(nums=nums)
```

```python
nums = [1, 2, 3, 4]
Solution().maximumProduct(nums=nums)
```

```python
nums = [-1, -2, -3]
Solution().maximumProduct(nums=nums)
```
