```python
from typing import List


class Solution:
  def minSwaps(self, nums: List[int]) -> int:
    n = len(nums)
    arr = []
    for i in range(n):
      arr.append((sum(map(int, str(nums[i]))), nums[i], i))
    arr.sort()
    visited = [False] * n
    swaps = 0
    for i in range(n):
      if visited[i] or i == arr[i][2]:
        visited[i] = True
        continue
      cycle_size = 0
      j = i
      while not visited[j]:
        visited[j] = True
        j = arr[j][2]
        cycle_size += 1
      swaps += cycle_size - 1
    return swaps
```

```python
nums = [37, 100]
Solution().minSwaps(nums=nums)
```

```python
nums = [22, 14, 33, 7]
Solution().minSwaps(nums=nums)
```
