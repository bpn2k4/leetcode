```python
class Solution:
  def countDaysTogether(self, arriveAlice: str, leaveAlice: str, arriveBob: str, leaveBob: str) -> int:
    days_of_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    alice_start = sum(days_of_month[:int(arriveAlice[0:2]) - 1]) + int(arriveAlice[3:5])
    alice_end = sum(days_of_month[:int(leaveAlice[0:2]) - 1]) + int(leaveAlice[3:5])
    bob_start = sum(days_of_month[:int(arriveBob[0:2]) - 1]) + int(arriveBob[3:5])
    bob_end = sum(days_of_month[:int(leaveBob[0:2]) - 1]) + int(leaveBob[3:5])
    overlap = max(0, min(alice_end, bob_end) - max(alice_start, bob_start) + 1)
    return overlap
```

```python
arriveAlice = "08-15"
leaveAlice = "08-18"
arriveBob = "08-16"
leaveBob = "08-19"
Solution().countDaysTogether(arriveAlice=arriveAlice, leaveAlice=leaveAlice, arriveBob=arriveBob, leaveBob=leaveBob)
```
