```python
from typing import List

class Solution:
  def stoneGame(self, piles: List[int]) -> bool:
    n = len(piles)
    dp = [0] * n
    for i in range(n - 1, -1, -1):
      dp[i] = piles[i]
      for j in range(i + 1, n):
        dp[j] = max(piles[i] - dp[j], piles[j] - dp[j - 1])
    return dp[n - 1] >= 0
  
  def stoneGame(self, piles: List[int]) -> bool:
    return True
```

```python
piles = [5, 3, 4, 5]
Solution().stoneGame(piles=piles)
```
