```python
from typing import List


class Solution:
  def maxAlternatingSum(self, nums: List[int]) -> int:
    n = len(nums)
    nums = sorted([num * num for num in nums], reverse=True)
    s = 0
    for i in range((n + 1) // 2):
      s += nums[i]
    for i in range((n + 1) // 2, n):
      s -= nums[i]
    return s
```

```python
nums = [1, 2, 3]
Solution().maxAlternatingSum(nums=nums)
```

```python
nums = [1, -1, 2, -2, 3, -3]
Solution().maxAlternatingSum(nums=nums)
```
