```python
from typing import List


class Solution:
  def minOperations(self, nums: List[int]) -> int:
    n = len(nums)
    if n == 1:
      return 0
    start = nums.index(0)
    asc = True
    desc = True
    for i in range(1, n):
      value = nums[(start + i) % n]
      if value != i:
        asc = False
      if value != n - i:
        desc = False
      if not asc and not desc:
        return -1
    if asc:
      return min(start, n - start + 2)
    return min(n - start, start + 2)
```

```python
nums = [0, 2, 1]
Solution().minOperations(nums=nums)
```

```python
nums = [1, 0, 2]
Solution().minOperations(nums=nums)
```

```python
nums = [2, 0, 1, 3]
Solution().minOperations(nums=nums)
```
