```python
from typing import List


class Solution:
  def stoneGameV(self, stoneValue: List[int]) -> int:
    n = len(stoneValue)
    f = [[0] * n for _ in range(n)]
    max_left = [[0] * n for _ in range(n)]
    max_right = [[0] * n for _ in range(n)]
    for left in range(n - 1, -1, -1):
      max_left[left][left] = max_right[left][left] = stoneValue[left]
      total = stoneValue[left]
      sum_left = 0
      i = left - 1
      for right in range(left + 1, n):
        total += stoneValue[right]
        while i + 1 < right and (sum_left + stoneValue[i + 1]) * 2 <= total:
          sum_left += stoneValue[i + 1]
          i += 1
        if left <= i:
          f[left][right] = max(f[left][right], max_left[left][i])
        if i + 1 < right:
          f[left][right] = max(f[left][right], max_right[i + 2][right])
        if sum_left * 2 == total:
          f[left][right] = max(f[left][right], max_right[i + 1][right])
        max_left[left][right] = max(max_left[left][right - 1], total + f[left][right])
        max_right[left][right] = max(max_right[left + 1][right], total + f[left][right])
    return f[0][n - 1]
```

```python
stoneValue = [6, 2, 3, 4, 5, 5]
Solution().stoneGameV(stoneValue=stoneValue)
```
