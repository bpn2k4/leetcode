```python
class Solution:
  def removeOuterParentheses(self, s: str) -> str:
    n = len(s)
    output = ''
    stack = ['(']
    p = ''
    for i in range(1, n):
      char = s[i]
      if char == '(':
        stack.append(char)
        if len(stack) > 1:
          p += char
      else:
        stack.pop()
        if len(stack) == 0:
          output += p
          p = ''
        else:
          p += char
    return output
```

```python
s = "(()())(())"
Solution().removeOuterParentheses(s=s)
```

```python
s = "(()())(())(()(()))"
Solution().removeOuterParentheses(s=s)
```
