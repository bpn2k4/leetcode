```python
class Solution:
  def minimumCost(self, cost1: int, cost2: int, costBoth: int, need1: int, need2: int) -> int:
    return min(
      need1 * cost1 + need2 * cost2,
      max(need1, need2) * costBoth,
      need1 * costBoth + max(0, need2 - need1) * cost2,
      need2 * costBoth + max(0, need1 - need2) * cost1,
    )
```

```python
cost1 = 3
cost2 = 2
costBoth = 1
need1 = 3
need2 = 2
Solution().minimumCost(cost1=cost1, cost2=cost2, costBoth=costBoth, need1=need1, need2=need2)
```
