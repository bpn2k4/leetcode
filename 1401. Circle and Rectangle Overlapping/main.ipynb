```python
class Solution:
  def checkOverlap(self, radius: int, xCenter: int, yCenter: int, x1: int, y1: int, x2: int, y2: int) -> bool:
    if x1 <= (xCenter - radius) and x2 >= (xCenter + radius) and y1 <= (yCenter - radius) and y2 >= (yCenter + radius):
      return True
    x3, y3 = x1, y2
    x4, y4 = x2, y1
    r2 = radius * radius
    if (
      (x1 - xCenter) * (x1 - xCenter) + (y1 - yCenter) * (y1 - yCenter) <= r2 or
      (x2 - xCenter) * (x2 - xCenter) + (y2 - yCenter) * (y2 - yCenter) <= r2 or
      (x3 - xCenter) * (x3 - xCenter) + (y3 - yCenter) * (y3 - yCenter) <= r2 or
      (x4 - xCenter) * (x4 - xCenter) + (y4 - yCenter) * (y4 - yCenter) <= r2
    ):
      return True
    m1 = (x1 - xCenter) * (x1 - xCenter)
    m2 = (x2 - xCenter) * (x2 - xCenter)
    if r2 >= m1:
      y_1 = yCenter - (r2 - m1) ** 0.5
      y_2 = yCenter + (r2 - m1) ** 0.5
      if y1 <= y_1 <= y2 or y1 <= y_2 <= y2:
        return True
    if r2 >= m2:
      y_3 = yCenter - (r2 - m2) ** 0.5
      y_4 = yCenter + (r2 - m2) ** 0.5
      if y1 <= y_3 <= y2 or y1 <= y_4 <= y2:
        return True
    n1 = (y1 - yCenter) * (y1 - yCenter)
    n2 = (y2 - yCenter) * (y2 - yCenter)
    if r2 >= n1:
      x_1 = xCenter - (r2 - n1) ** 0.5
      x_2 = xCenter + (r2 - n1) ** 0.5
      if x1 <= x_1 <= x2 or x1 <= x_2 <= x2:
        return True
    if r2 >= n2:
      x_3 = xCenter - (r2 - n2) ** 0.5
      x_4 = xCenter + (r2 - n2) ** 0.5
      if x1 <= x_3 <= x2 or x1 <= x_4 <= x2:
        return True
    return False
```
