```python
from typing import List


class Solution:
  def findMissingElements(self, nums: List[int]) -> List[int]:
    min_num = min(nums)
    max_num = max(nums)
    mark = {num: 1 for num in nums}
    missing = []
    for num in range(min_num + 1, max_num):
      if num not in mark:
        missing.append(num)
    return missing
```

```python
nums = [1, 4, 2, 5]
Solution().findMissingElements(nums=nums)
```

```python
nums = [7, 8, 6, 9]
Solution().findMissingElements(nums=nums)
```

```python
nums = [5, 1]
Solution().findMissingElements(nums=nums)
```
