```python
class Solution:
  def generateParenthesis(self, n: int) -> list[str]:
    output = []

    def back_tracking(l: int, r: int, s: str):
      if l == n:
        if r == n:
          output.append(s)
        else:
          back_tracking(l, r + 1, s + ')')
      elif l == r:
        back_tracking(l + 1, r, s + '(')
      else:
        back_tracking(l + 1, r, s + '(')
        back_tracking(l, r + 1, s + ')')

    back_tracking(0, 0, '')
    return output
```

```python
n = 3
Solution().generateParenthesis(n=n)
```

```python
n = 1
Solution().generateParenthesis(n=n)
```
