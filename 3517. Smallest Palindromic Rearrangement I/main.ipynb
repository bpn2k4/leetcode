```python
from collections import Counter


class Solution:
  def smallestPalindrome(self, s: str) -> str:
    char2count = Counter(s)
    char2count = [(char, count) for char, count in char2count.items()]
    char2count.sort()
    if len(s) % 2 == 0:
      s = ''
      for char, count in char2count:
        s += char * (count // 2)
      s += s[::-1]
      return s
    else:
      s = ''
      mid = ''
      for char, count in char2count:
        if count % 2 == 1:
          mid = char
        s += char * (count // 2)
      s += mid + s[::-1]
      return s
```

```python
s = "z"
Solution().smallestPalindrome(s=s)
```

```python
s = "babab"
Solution().smallestPalindrome(s=s)
```
