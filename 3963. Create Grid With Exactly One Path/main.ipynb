```python
class Solution:
  def createGrid(self, m: int, n: int) -> list[str]:
    grid = ['.' + '#' * (n - 1) for _ in range(m - 1)]
    grid.append('.' * n)
    return grid
```

```python
m = 2
n = 3
Solution().createGrid(m=m, n=n)
```
