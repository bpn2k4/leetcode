```python
class Solution:
  def addStrings(self, num1: str, num2: str) -> str:
    n1, n2 = len(num1), len(num2)
    if n1 > n2:
      num2 = '0' * (n1 - n2) + num2
    elif n2 > n1:
      num1 = '0' * (n2 - n1) + num1
    n = len(num1)
    ans = ''
    c = 0
    for i in range(n - 1, -1, -1):
      num = int(num1[i]) + int(num2[i]) + c
      ans = str(num % 10) + ans
      c = num // 10
    if c:
      ans = str(c) + ans
    return ans
```

```python
num1 = "11"
num2 = "123"
Solution().addStrings(num1=num1, num2=num2)
```
